```python
import uproot
```

```python
filter_name = [
    "showers_size", 
    "showers_nDigis", 
    # "showers_nShowers_*", 
    # "realshowers_nDigis_*", 
    # "realshowers_size_*", 
    # "realshowers_nShowers_*", 
    "showered_genmuon_meth", 
    "shower_tpfptnfn",
    "showers_classification",
    # "shower_filter_highpt_tag_eff_* "
]
```

```python
G_org = uproot.open("histograms_zp_G.root")
```

```python
filtered_histos = {}
for key in G_org.keys():
    if any([f in key for f in filter_name]):
        filtered_histos[key.replace(";1", "")] = G_org[key]
```

```python
filter_name = [
    "realshowers",
    "shower_filter_highpt_tag_eff"
]
```

```python
G_news = uproot.open("histograms_zp_G_news.root")
```

```python
filtered_histos_Gnews = {}
for key in G_news.keys():
    if any([f in key for f in filter_name]):
        filtered_histos_Gnews[key.replace(";1", "")] = G_news[key]
```

```python
G_news2 = uproot.open("histograms_zp_G_news2.root")
```

```python
filtered_histos_Gnews2 = {}
for key in G_news2.keys():
    if "nShowers" in key:
        filtered_histos_Gnews2["showers_" + key.replace(";1", "")] = G_news2[key]
```

```python
with uproot.recreate("../histograms_zp_G.root") as f:
    for key, histo in filtered_histos.items():
        f[key] = histo
    for key, histo in filtered_histos_Gnews.items():
        f[key] = histo
    for key, histo in filtered_histos_Gnews2.items():
        f[key] = histo
```
